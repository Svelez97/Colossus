```python
import tkinter as tk
from tkinter import Button, Label, Text, Entry
from tkinter import ttk
from PIL import Image, ImageTk
from tkinter import filedialog, messagebox
import pandas as pd
from IPython.display import clear_output
import os
import re

CORE_folder = str("/".join(str(os.getcwd()).replace("\\","/").split("/")[:]))
print(CORE_folder)

# Crear la ventana principal
root = tk.Tk()
root.title("Massive Files Filter - C.O.R.E")
root.geometry("1530x660+0+80")
root.iconbitmap("images/LOGO_total.ico")

# select_file_bg
select_file_bg = Image.open(CORE_folder+"/images/select_file.png")  # Reemplaza con la ruta a tu imagen
select_file_bg = ImageTk.PhotoImage(select_file_bg)

# new_file
new_file = Image.open(CORE_folder+"/images/new_file.png")  # Reemplaza con la ruta a tu imagen
new_file = ImageTk.PhotoImage(new_file)

# titles_filter
titles = Image.open(CORE_folder+"/images/titles.png")  # Reemplaza con la ruta a tu imagen
titles = ImageTk.PhotoImage(titles)

# yes_icon
yes = Image.open(CORE_folder+"/images/yes.png")  # Reemplaza con la ruta a tu imagen
yes = ImageTk.PhotoImage(yes)

# no_icon
no = Image.open(CORE_folder+"/images/no.png")  # Reemplaza con la ruta a tu imagen
no = ImageTk.PhotoImage(no)

# lens_icon
lens = Image.open(CORE_folder+"/images/lens.png")  # Reemplaza con la ruta a tu imagen
lens = ImageTk.PhotoImage(lens)

# preview_icon
preview_i = Image.open(CORE_folder+"/images/preview.png")  # Reemplaza con la ruta a tu imagen
preview_i = ImageTk.PhotoImage(preview_i)

# export_icon
export_i = Image.open(CORE_folder+"/images/export.png")  # Reemplaza con la ruta a tu imagen
export_i = ImageTk.PhotoImage(export_i)

# col_1_filter_icon
col_1_filter = Image.open(CORE_folder+"/images/col_1_filter.png")  # Reemplaza con la ruta a tu imagen
col_1_filter = ImageTk.PhotoImage(col_1_filter)

# SW_filter_icon
SW_filter_icon = Image.open(CORE_folder+"/images/SW_filter.png")  # Reemplaza con la ruta a tu imagen
SW_filter_icon = ImageTk.PhotoImage(SW_filter_icon)

# SW_stats_icon
SW_stats_icon = Image.open(CORE_folder+"/images/SW_stats.png")  # Reemplaza con la ruta a tu imagen
SW_stats_icon = ImageTk.PhotoImage(SW_stats_icon)

# min_max_icon
min_max_icon = Image.open(CORE_folder+"/images/min_max.png")  # Reemplaza con la ruta a tu imagen
min_max_icon = ImageTk.PhotoImage(min_max_icon)

# Función para borrar todo el contenido de la ventana
def clear_window():
    for widget in root.winfo_children():
        widget.destroy()

def pop_up_message(message, close_time = None):
    messagebox.showinfo("CORE System Message", message)

def detect_separator(file_csv):
    with open(file_csv, 'r') as file:
        line_1 = file.readline()
        if '|' in line_1:
            return '|'
        elif ',' in line_1:
            return ','

def select_file2():
    f1 = filedialog.askopenfilename(initialdir= "", title="Select your CSV file")
    if f1 != "":
        separator = detect_separator(f1)
        df = pd.read_csv(f1, sep= separator, nrows=5)
        columns_to_filter = list(df.columns.values)
        MFF_view(f1, columns_to_filter)

from tkinter import simpledialog
def get_data(question):
    answer = simpledialog.askstring("CORE-Box Entry", question)
    return answer

############################################################################################################################################################################
############################################################################################################################################################################
########################################################################### MAIN WINDOW CREATION ###########################################################################
############################################################################################################################################################################
############################################################################################################################################################################

def MFF_view(f1 = "", df_columns= [""]):

    clear_window()

    # file bar
    TITLE_1 = Label(root, text = "File: ", compound="center", font=("Arial", 15, "bold"))
    TITLE_1.place(x = 10, y = 10)

    TITLE_1 = Label(root, text = "File to export: ", compound="center", font=("Arial", 15, "bold"))
    TITLE_1.place(x = 560, y = 10)

    f1_name = str("/".join(str(f1).replace("\\","/").split("/")[-1:]))
    f1_name = re.sub(r".csv", "", f1_name)

    if f1 == "":
        file_path = Label(root, text="Select your massive file...", image=select_file_bg, compound='center', borderwidth=0, highlightthickness=0, cursor="hand2", font=("Arial", 11, "bold"), width=460)
        file_path.place(x=60, y= 10)
        file_path.bind("<Button-1>", lambda e: select_file2())

        # file to export
        filter_file = Label(root, text="", image=new_file, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
        filter_file.place(x=705, y= 10)

    else:

        file_path = Label(root, text=str("/".join(str(f1).replace("\\","/").split("/")[-1:])), image=select_file_bg, compound='center', borderwidth=0, highlightthickness=0, cursor="hand2", font=("Arial", 11, "bold"), width=460)
        file_path.place(x=60, y= 10)
        file_path.bind("<Button-1>", lambda e: select_file2())

        # file to export
        filter_file_L = Label(root, image=new_file, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
        filter_file_L.place(x=705, y= 10)

        # file to export
        filter_file = Entry(root, width=38, font=("Arial", 11, "bold"), bg = "#F6C6AD", bd=0, highlightthickness=0)
        filter_file.insert(0, f1_name + "_FILTER")
        filter_file.place(x = 720, y = 17)

    ###################### filter space ######################

    def buscar(event=None):
        # Limpiar resultados anteriores
        U_Values = column_val_CB['values']
        column_val_CB['values'] = []
        resultados = []
        
        # Obtener texto de búsqueda
        texto_busqueda = entrada_busqueda.get().lower()

        # Filtrar nombres
        for nombre in U_Values:
            if texto_busqueda in nombre.lower():
                resultados.append(nombre)
        
        # Actualizar combobox con resultados
        column_val_CB['values'] = resultados
        if texto_busqueda:
            pass
        else:
            column_val_CB['values'] = U_Values

    def seleccionar(event):
        nombre_seleccionado = column_val_CB.get()
        entrada_busqueda.delete(0, tk.END)
        entrada_busqueda.insert(0, nombre_seleccionado)

    def select_col_1(file):

        path = str("/".join(str(file).replace("\\","/").split("/")[:-1]))

        f1_name = str("/".join(str(file).replace("\\","/").split("/")[-1:]))
        f1_name = re.sub(r".csv", "", f1_name)

        import polars as pl
        sep = detect_separator(path + "/" + f1_name + ".csv")
        df_v = pl.read_csv(path + "/" + f1_name + ".csv", separator=sep, columns=[column_fil.get()], infer_schema_length=10000)

        unique_val = df_v.select(column_fil.get()).unique().to_series().to_list()

        if len(unique_val) <= 900:
            try:
                unique_val = sorted(unique_val)
            except:
                pass
            column_val_CB.configure(values = unique_val)
        else:
            column_val_CB.configure(values = ["Unique values out of range!!!"])


    tall_button = Label(root, image=col_1_filter, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=1310, y=57)

    # Unique values
    column_fil = ttk.Combobox(root, values=df_columns, width=28, justify="center", height=30)
    column_fil.set(df_columns[-1])
    column_fil.place(x = 1319, y = 62)
    column_fil.bind('<<ComboboxSelected>>', lambda e: select_col_1(f1))

    column_val_CB = ttk.Combobox(root, values=[""], width=28, justify="center")
    column_val_CB.set("")
    column_val_CB.place(x = 1319, y = 87)
    column_val_CB.bind('<<ComboboxSelected>>', seleccionar)

    # Crear entrada de búsqueda
    entrada_busqueda = Entry(root, width=28, justify='left')
    entrada_busqueda.place(x=1319, y=87+0)
    entrada_busqueda.bind('<KeyRelease>', buscar)

    # simbol
    simbol_list = ["in", "not in", "=", "!=", ">", ">=", "<", "<=", "Like", "Not Like"]

    borderY = 30
    borderX = -300

    ########## filter 1
    borderX += 300

    tall_button = Label(root, image=titles, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=14 + borderX, y=27 + borderY)
    
    column_fil_1 = ttk.Combobox(root, values=df_columns, width=23, justify="center", height=30)
    column_fil_1.set("")
    column_fil_1.place(x = 14 + borderX, y = 57 + borderY)

    column_simb_1 = ttk.Combobox(root, values=simbol_list, width=10, justify="center")
    column_simb_1.set("")
    column_simb_1.place(x = 182 + borderX, y = 57 + borderY)    

    # values 1
    column_val_1 = Entry(root, width=41, borderwidth=0, highlightthickness=1)
    column_val_1.place(x = 15 + borderX, y = 82 + borderY)

    ########## filter 2
    borderX += 260
    tall_button = Label(root, image=titles, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=14 + borderX, y=27 + borderY)
    
    column_fil_2 = ttk.Combobox(root, values=df_columns, width=23, justify="center", height=30)
    column_fil_2.set("")
    column_fil_2.place(x = 14 + borderX, y = 57 + borderY)

    column_simb_2 = ttk.Combobox(root, values=simbol_list, width=10, justify="center")
    column_simb_2.set("")
    column_simb_2.place(x = 182 + borderX, y = 57 + borderY)

    # values 2
    column_val_2 = Entry(root, width=41, borderwidth=0, highlightthickness=1)
    column_val_2.place(x = 14 + borderX, y = 82 + borderY)

    ########## filter 3
    borderX += 260
    tall_button = Label(root, image=titles, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=14 + borderX, y=27 + borderY)
    
    column_fil_3 = ttk.Combobox(root, values=df_columns, width=23, justify="center", height=30)
    column_fil_3.set("")
    column_fil_3.place(x = 14 + borderX, y = 57 + borderY)

    column_simb_3 = ttk.Combobox(root, values=simbol_list, width=10, justify="center")
    column_simb_3.set("")
    column_simb_3.place(x = 182 + borderX, y = 57 + borderY)

    # values 3
    column_val_3 = Entry(root, width=41, borderwidth=0, highlightthickness=1)
    column_val_3.place(x = 14 + borderX, y = 82 + borderY)

    ########## filter 4
    borderX += 260
    tall_button = Label(root, image=titles, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=14 + borderX, y=27 + borderY)
    
    column_fil_4 = ttk.Combobox(root, values=df_columns, width=23, justify="center", height=30)
    column_fil_4.set("")
    column_fil_4.place(x = 14 + borderX, y = 57 + borderY)

    column_simb_4 = ttk.Combobox(root, values=simbol_list, width=10, justify="center")
    column_simb_4.set("")
    column_simb_4.place(x = 182 + borderX, y = 57 + borderY)

    # values 4
    column_val_4 = Entry(root, width=41, borderwidth=0, highlightthickness=1)
    column_val_4.place(x = 14 + borderX, y = 82 + borderY)

    ########## filter 5
    borderX += 260
    tall_button = Label(root, image=titles, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"))
    tall_button.place(x=14 + borderX, y=27 + borderY)
    
    column_fil_5 = ttk.Combobox(root, values=df_columns, width=23, justify="center", height=30)
    column_fil_5.set("")
    column_fil_5.place(x = 14 + borderX, y = 57 + borderY)

    column_simb_5 = ttk.Combobox(root, values=simbol_list, width=10, justify="center")
    column_simb_5.set("")
    column_simb_5.place(x = 182 + borderX, y = 57 + borderY)

    # values 5
    column_val_5 = Entry(root, width=41, borderwidth=0, highlightthickness=1)
    column_val_5.place(x = 14 + borderX, y = 82 + borderY)

    # negate the filters
    TITLE_2 = Label(root, text = "Negate the filters:", compound="center", font=("Arial", 14, "bold"))
    TITLE_2.place(x = 10, y = 115 + borderY)

    window_list_3 = ["No", "Yes"]
    column_simb_6 = ttk.Combobox(root, values=window_list_3, width=5, justify="center")
    column_simb_6.set("No")
    column_simb_6.place(x = 180, y = 145)

    def negate_filter(value):
        if value == "No":
            negate_button.configure(image=yes)
            column_simb_6.set("Yes")

        elif value == "Yes":
            negate_button.configure(image=no)
            column_simb_6.set("No")

    negate_button = Label(root, image=no, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"), cursor="hand2")
    negate_button.bind("<Button-1>", lambda e: negate_filter(column_simb_6.get()))
    negate_button.place(x=180, y=145)

    preview_button = Label(root, image=preview_i, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"), cursor="hand2")
    preview_button.bind("<Button-1>", lambda e: mini_preview(f1,
                                                            column_fil_1.get(),
                                                            column_simb_1.get(),
                                                            column_val_1.get(),
                                                            column_fil_2.get(),
                                                            column_simb_2.get(),
                                                            column_val_2.get(),
                                                            column_fil_3.get(),
                                                            column_simb_3.get(),
                                                            column_val_3.get(),
                                                            column_fil_4.get(),
                                                            column_simb_4.get(),
                                                            column_val_4.get(),
                                                            column_fil_5.get(),
                                                            column_simb_5.get(),
                                                            column_val_5.get(),
                                                            column_simb_6.get(), root))
    preview_button.place(x = 435, y = 115 + borderY)

    min_max_button = Label(root, image=min_max_icon, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"), cursor="hand2")
    min_max_button.bind("<Button-1>", lambda e: mini_preview_2(f1,
                                                                column_fil_1.get(),
                                                                column_simb_1.get(),
                                                                column_val_1.get(),
                                                                column_fil_2.get(),
                                                                column_simb_2.get(),
                                                                column_val_2.get(),
                                                                column_fil_3.get(),
                                                                column_simb_3.get(),
                                                                column_val_3.get(),
                                                                column_fil_4.get(),
                                                                column_simb_4.get(),
                                                                column_val_4.get(),
                                                                column_fil_5.get(),
                                                                column_simb_5.get(),
                                                                column_val_5.get(),
                                                                column_simb_6.get(),
                                                                column_fil.get()))
    min_max_button.place(x = 570, y = 115 + borderY)

    export_button = Label(root, image=export_i, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"), cursor="hand2")
    export_button.bind("<Button-1>", lambda e: deploy_Massive_files_filter(f1,
                                                                            column_fil_1.get(),
                                                                            column_simb_1.get(),
                                                                            column_val_1.get(),
                                                                            column_fil_2.get(),
                                                                            column_simb_2.get(),
                                                                            column_val_2.get(),
                                                                            column_fil_3.get(),
                                                                            column_simb_3.get(),
                                                                            column_val_3.get(),
                                                                            column_fil_4.get(),
                                                                            column_simb_4.get(),
                                                                            column_val_4.get(),
                                                                            column_fil_5.get(),
                                                                            column_simb_5.get(),
                                                                            column_val_5.get(),
                                                                            column_simb_6.get(),
                                                                            filter_file.get()))
    export_button.place(x = 300, y = 115 + borderY)

    root.update_idletasks()
    
    # Crear un cuadro de texto grande
    text_box = Text(root, wrap="word", width=150, height=16, font=("Arial", 12), borderwidth=0, highlightthickness=0, bg="lightgray")
    text_box.place(x=75, y=350)

    # configure the tags
    text_box.tag_configure("bold", font=("Arial", 12, "bold"))

    # Texto sobre Star Wars
    instructions = """CODE INSTRUCTIONS
Context: This code was created to filter files that are too large to be opened using Excel. The UI allows you to filter a file using up to five filters at once, and you can also choose to negate the filters.

1. Select the file that will be filtered. (The code can read files separated by "," and "|".)
2. Set the name of the file to export (don't include the .csv).
3. Set the filter columns to use.
4. Set the criteria for the filter.
    - Note: If you want to filter a list of values, separte the list with a comma:
            ex1: 1,2,3,4,5,6
            ex2: TOTAL US DOLLAR, TOTAL US DRUG, TOTAL US FOOD

5. It is not necessary to use all the filters. Just select what you need for each filter and leave the others blank.
    - Note: You can preview the filtered data before export it.
    - Click on the "magnifying glass icon" to check the filter data, the amount of data that can be preview is up to 200 but you can change the bucket.
6. Click on EXPORT CSV to deploy the filtering process and check the output file in the same path."""

    # Insertar el texto en el cuadro de texto
    text_box.insert("1.0", instructions)

    # Aplicar el estilo de negrita a las partes correspondientes
    text_box.tag_add("bold", "1.0", "1.17")  # CODE INSTRUCTIONS
    text_box.tag_add("bold", "2.0", "2.8")  # Context

    text_box.configure(state="disabled")

    clear_output()
    


def condition_creation(dff, Column, ops, values):
        import re
        import polars as pl
        d_type = str(dff[Column].dtype)

        desire_values = list(re.split(r',\s*', str(values)))
        desire_values = [value.strip() for value in desire_values]
        try:
            if d_type == "Int64":
                desire_values = [int(value) for value in desire_values]
            elif d_type == "Float64":
                desire_values = [float(value) for value in desire_values]
            elif d_type == "String":
                desire_values = [str(value) for value in desire_values]
        
        except:
            desire_values = [str(value) for value in desire_values]

        if ops == "in" or ops == "not in" or len(desire_values)>=2:
            if ops == "in" or ops == "=":
                cond_n = dff[Column].is_in(desire_values)
            elif ops == "not in" or ops == "!=":
                cond_n = ~(dff[Column].is_in(desire_values))

        elif ops == "=":
            cond_n = dff[Column] == desire_values[0]
        elif ops == "!=":
            cond_n = dff[Column] != desire_values[0]
        elif ops == ">":
            cond_n = dff[Column] > desire_values[0]
        elif ops == ">=":
            cond_n = dff[Column] >= desire_values[0]
        elif ops == "<":
            cond_n = dff[Column] < desire_values[0]
        elif ops == "<=":
            cond_n = dff[Column] <= desire_values[0]
        elif ops == "Like":
            dff = dff.with_columns(pl.col(Column).str.to_lowercase().alias("column_lower"))
            cond_n = dff['column_lower'].str.contains(desire_values[0].lower())
        elif ops == "Not Like":
            dff = dff.with_columns(pl.col(Column).str.to_lowercase().alias("column_lower"))
            cond_n = ~dff['column_lower'].str.contains(desire_values[0].lower())

        del dff
        return cond_n

def deploy_Massive_files_filter(file_path, column_fil_1, column_simb_1, column_val_1, column_fil_2, column_simb_2, column_val_2, column_fil_3, column_simb_3, column_val_3, column_fil_4, column_simb_4, column_val_4, column_fil_5, column_simb_5, column_val_5, column_simb_6, filter_file):
    import os
    import re
    
    path = str("/".join(str(file_path).replace("\\","/").split("/")[:-1]))

    f1_name = str("/".join(str(file_path).replace("\\","/").split("/")[-1:]))
    f1_name = re.sub(r".csv", "", f1_name)
    
    CORE_path = os.getcwd()
    CORE_path = str("/".join(str(CORE_path).replace("\\","/").split("/")[:-1]))

    try:
        # Lee el archivo CSV
        import polars as pl
        sep = detect_separator(path + "/" + f1_name + ".csv")
        df = pl.read_csv(path + "/" + f1_name + ".csv", separator=sep, infer_schema_length=10000)
        original_len = f"{df.height:,}"
        column_list = df.columns

        true_table = pl.DataFrame()

        if str(column_val_1) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_1, column_simb_1, column_val_1).alias("Bool_value_1"))

        if str(column_val_2) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_2, column_simb_2, column_val_2).alias("Bool_value_2"))

        if str(column_val_3) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_3, column_simb_3, column_val_3).alias("Bool_value_3"))

        if str(column_val_4) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_4, column_simb_4, column_val_4).alias("Bool_value_4"))

        if str(column_val_5) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_5, column_simb_5, column_val_5).alias("Bool_value_5"))
        
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        
        column_names = true_table.columns

        True_column = true_table[column_names[0]]
        for col in column_names:
            True_column = True_column & true_table[col]

        del true_table

        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########

        # Aplicación de la lógica
        if column_simb_6 == "No":
            df = df.filter(True_column)
        elif column_simb_6 == "Yes":
            df = df.filter(~True_column)

        del True_column
        
        df = df.select(column_list)

        new_file_name = path + f"/{filter_file}.csv"
        new_len = f"{df.height:,}"
        
        df.write_csv(new_file_name, separator=sep, line_terminator='\n')

        pop_up_message(f"{f1_name} file updates:\n\n     - Initial # of rows: {original_len} rows.\n     - Final # of rows: {new_len} rows.")
        pop_up_message("Process Completed")

        del df

    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

def mini_preview(file_path, column_fil_1, column_simb_1, column_val_1, column_fil_2, column_simb_2, column_val_2, column_fil_3, column_simb_3, column_val_3, column_fil_4, column_simb_4, column_val_4, column_fil_5, column_simb_5, column_val_5, column_simb_6, root):

    import os
    import re
    global preview_W
    
    path = str("/".join(str(file_path).replace("\\","/").split("/")[:-1]))

    f1_name = str("/".join(str(file_path).replace("\\","/").split("/")[-1:]))
    f1_name = re.sub(r".csv", "", f1_name)

    # Lee el archivo CSV
    try:
        import polars as pl
        sep = detect_separator(path + "/" + f1_name + ".csv")
        df = pl.read_csv(path + "/" + f1_name + ".csv", separator=sep, infer_schema_length=10000)
        column_list = df.columns

        true_table = pl.DataFrame()

        if str(column_val_1) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_1, column_simb_1, column_val_1).alias("Bool_value_1"))

        if str(column_val_2) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_2, column_simb_2, column_val_2).alias("Bool_value_2"))

        if str(column_val_3) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_3, column_simb_3, column_val_3).alias("Bool_value_3"))

        if str(column_val_4) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_4, column_simb_4, column_val_4).alias("Bool_value_4"))

        if str(column_val_5) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_5, column_simb_5, column_val_5).alias("Bool_value_5"))
        
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        
        column_names = true_table.columns

        True_column = true_table[column_names[0]]
        for col in column_names:
            True_column = True_column & true_table[col]

        del true_table

        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########

        # Aplicación de la lógica
        if column_simb_6 == "No":
            df = df.filter(True_column)
        elif column_simb_6 == "Yes":
            df = df.filter(~True_column)

        del True_column
        
        df = df.select(column_list)

    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

    try:  
        df = df.slice(0, 6)

        # Crear un frame dentro de la ventana principal con tamaño 1000 x 200
        frame = tk.Frame(root, width=1400, height=140)
        frame.place(x=20, y=195)  # Definir la posición del frame en la ventana

        # Crear un canvas para agregar barras de desplazamiento
        canvas = tk.Canvas(frame, width=1400, height=140)
        canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

        # Agregar barras de desplazamiento al canvas
        scroll_x = ttk.Scrollbar(frame, orient="horizontal", command=canvas.xview)
        scroll_y = ttk.Scrollbar(frame, orient="vertical", command=canvas.yview)
        canvas.configure(xscrollcommand=scroll_x.set, yscrollcommand=scroll_y.set)

        scroll_x.pack(side="bottom", fill="x")
        # scroll_y.pack(side="right", fill="y")

        # Crear otro frame dentro del canvas
        inner_frame = tk.Frame(canvas)
        canvas.create_window((0, 0), window=inner_frame, anchor="nw")

        # Crear un Treeview para mostrar el dataframe
        tree = ttk.Treeview(inner_frame)

        # Definir las columnas del Treeview
        tree["columns"] = list(df.columns)
        tree["show"] = "headings"

        # Configurar los encabezados de las columnas
        for column in df.columns:
            tree.heading(column, text=column)

        # Agregar las filas del dataframe al Treeview
        for row in df.iter_rows():
            tree.insert("", "end", values=list(row))

        # Empacar el Treeview en el inner_frame
        tree.pack()

        # Ajustar el tamaño del inner_frame para que se ajuste al contenido
        inner_frame.update_idletasks()
        canvas.config(scrollregion=canvas.bbox("all"))

        lens_button = Label(root, image=lens, compound='center', borderwidth=0, highlightthickness=0, font=("Arial", 11, "bold"), cursor="hand2")
        lens_button.bind("<Button-1>", lambda e: preview(file_path, column_fil_1, column_simb_1, column_val_1, column_fil_2, column_simb_2, column_val_2, column_fil_3, column_simb_3, column_val_3, column_fil_4, column_simb_4, column_val_4, column_fil_5, column_simb_5, column_val_5, column_simb_6))
        lens_button.place(x=1440, y=200)


    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

def mini_preview_2(file_path, column_fil_1, column_simb_1, column_val_1, column_fil_2, column_simb_2, column_val_2, column_fil_3, column_simb_3, column_val_3, column_fil_4, column_simb_4, column_val_4, column_fil_5, column_simb_5, column_val_5, column_simb_6, min_max_col):

    import os
    import re
    global preview_W
    
    path = str("/".join(str(file_path).replace("\\","/").split("/")[:-1]))

    f1_name = str("/".join(str(file_path).replace("\\","/").split("/")[-1:]))
    f1_name = re.sub(r".csv", "", f1_name)

    # Lee el archivo CSV
    try:
        import polars as pl
        sep = detect_separator(path + "/" + f1_name + ".csv")
        df = pl.read_csv(path + "/" + f1_name + ".csv", separator=sep, infer_schema_length=10000)
        column_list = df.columns

        true_table = pl.DataFrame()

        if str(column_val_1) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_1, column_simb_1, column_val_1).alias("Bool_value_1"))

        if str(column_val_2) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_2, column_simb_2, column_val_2).alias("Bool_value_2"))

        if str(column_val_3) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_3, column_simb_3, column_val_3).alias("Bool_value_3"))

        if str(column_val_4) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_4, column_simb_4, column_val_4).alias("Bool_value_4"))

        if str(column_val_5) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_5, column_simb_5, column_val_5).alias("Bool_value_5"))
        
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        
        column_names = true_table.columns

        True_column = true_table[column_names[0]]
        for col in column_names:
            True_column = True_column & true_table[col]

        del true_table

        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########

        # Aplicación de la lógica
        if column_simb_6 == "No":
            df = df.filter(True_column)
        elif column_simb_6 == "Yes":
            df = df.filter(~True_column)

        del True_column
        
        df = df.select(column_list)

    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

    try:
        unique_val = df.select(min_max_col).unique().to_series().to_list()
        try:
                unique_val = sorted(unique_val)
        except:
            pass
        pop_up_message(f'Column Selected: "{min_max_col}"\n\n Min Value: {unique_val[0]}\n Max Value: {unique_val[-1]}')

    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

def canvas_for_preview():
    import tkinter as tk
    from tkinter import ttk 
    # Create A Main frame
    Preview_result = tk.Tk()
    Preview_result.geometry("500x300+300+10")
    
    main_frame = tk.Frame(Preview_result)
    main_frame.pack(fill=tk.BOTH,expand=1)

    # Create Frame for X Scrollbar
    sec = tk.Frame(main_frame)
    sec.pack(fill=tk.X,side=tk.BOTTOM)

    # Create A Canvas
    my_canvas = tk.Canvas(main_frame)
    my_canvas.pack(side=tk.LEFT,fill=tk.BOTH,expand=1)

    # Add A Scrollbars to Canvas
    x_scrollbar = ttk.Scrollbar(sec,orient=tk.HORIZONTAL,command=my_canvas.xview)
    x_scrollbar.pack(side=tk.BOTTOM,fill=tk.X)
    
    y_scrollbar = ttk.Scrollbar(main_frame,orient=tk.VERTICAL,command=my_canvas.yview)
    y_scrollbar.pack(side=tk.RIGHT,fill=tk.Y)

    # Configure the canvas
    my_canvas.configure(xscrollcommand=x_scrollbar.set)
    my_canvas.configure(yscrollcommand=y_scrollbar.set)
    my_canvas.bind("<Configure>",lambda e: my_canvas.config(scrollregion= my_canvas.bbox(tk.ALL))) 

    # Create Another Frame INSIDE the Canvas
    second_frame = tk.Frame(my_canvas)

    # Add that New Frame a Window In The Canvas
    my_canvas.create_window((0,0),window= second_frame, anchor="nw")
    
    return Preview_result, second_frame

def preview(file_path, column_fil_1, column_simb_1, column_val_1, column_fil_2, column_simb_2, column_val_2, column_fil_3, column_simb_3, column_val_3, column_fil_4, column_simb_4, column_val_4, column_fil_5, column_simb_5, column_val_5, column_simb_6):

    import os
    import re
    global preview_W
    
    path = str("/".join(str(file_path).replace("\\","/").split("/")[:-1]))

    f1_name = str("/".join(str(file_path).replace("\\","/").split("/")[-1:]))
    f1_name = re.sub(r".csv", "", f1_name)

    # Lee el archivo CSV
    try:
        import polars as pl
        sep = detect_separator(path + "/" + f1_name + ".csv")
        df = pl.read_csv(path + "/" + f1_name + ".csv", separator=sep, infer_schema_length=10000)
        column_list = df.columns

        true_table = pl.DataFrame()

        if str(column_val_1) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_1, column_simb_1, column_val_1).alias("Bool_value_1"))

        if str(column_val_2) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_2, column_simb_2, column_val_2).alias("Bool_value_2"))

        if str(column_val_3) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_3, column_simb_3, column_val_3).alias("Bool_value_3"))

        if str(column_val_4) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_4, column_simb_4, column_val_4).alias("Bool_value_4"))

        if str(column_val_5) != "":
            true_table = true_table.with_columns(condition_creation(df, column_fil_5, column_simb_5, column_val_5).alias("Bool_value_5"))
        
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        
        column_names = true_table.columns

        True_column = true_table[column_names[0]]
        for col in column_names:
            True_column = True_column & true_table[col]

        del true_table

        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########
        # ########## MINI CHECK ##########

        # Aplicación de la lógica
        if column_simb_6 == "No":
            df = df.filter(True_column)
        elif column_simb_6 == "Yes":
            df = df.filter(~True_column)

        del True_column
        
        df = df.select(column_list)

        df = df.with_row_index("ROW_ID")
    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

    try:
        preview_W.destroy()
    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

    try:
        limit = 200
        n_R = f"{df.height:,}"

        ms = f"The filtered data has {n_R} rows.\nData preview limit is 200 rows.\n\nPreview the Data since row: "
        row_limit = int(get_data(ms))

        df = df.slice(row_limit, limit)

        column_names = df.columns
        preview_W, second_frame = canvas_for_preview()

        height = df.height
        width = len(column_names)

        for j in range(width):  # Columns
            globals()["column" + str(j)] = tk.StringVar()
            globals()["Entrycolumn" + str(j)] = tk.Entry(second_frame, textvariable=globals()["column" + str(j)], fg="black", bg="white")
            globals()["Entrycolumn" + str(j)].insert(0, column_names[j])
            globals()["Entrycolumn" + str(j)].grid(row=0, column=j)
            globals()["Entrycolumn" + str(j)].config(state='readonly')

        for i in range(height):  # Rows
            for j in range(width):  # Columns
                globals()["value" + str(i) + "_" + str(j)] = tk.StringVar()
                globals()["Entryvalue" + str(i) + "_" + str(j)] = tk.Entry(second_frame, textvariable=globals()["value" + str(i) + "_" + str(j)], fg="black", bg="white")
                globals()["Entryvalue" + str(i) + "_" + str(j)].insert(0, str(df.item(i, column_names[j])))
                globals()["Entryvalue" + str(i) + "_" + str(j)].grid(row=i+1, column=j)  # Ajuste en la fila
                globals()["Entryvalue" + str(i) + "_" + str(j)].config(state='readonly')
        
        del df

    except Exception as e:
        print(f"Tipo de error: {type(e).__name__}")

MFF_view()
root.mainloop()
```
